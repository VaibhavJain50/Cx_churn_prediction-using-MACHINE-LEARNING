```python
df.drop(['RowNumber','CustomerId','Surname'],axis=1, inplace=True)
```

```python
df=df.dropna()
```

```python
df['Age']=df['Age'].astype('int64')
```

```python
df['HasCrCard']=df['HasCrCard'].astype('int64')
df['IsActiveMember']=df['IsActiveMember'].astype('int64')
df['EstimatedSalary']=df['EstimatedSalary'].astype('int64')
```

```python
df=pd.get_dummies(df,columns=['Gender'], prefix='Is',drop_first=True)
```

```python
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
```

```python
df['Geography'] = label_encoder.fit_transform(df['Geography'])
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
df[['scaled_Balance','scaled_EstimatedSalary','scaled_CreditScore','scaled_Age']]=scaler.fit_transform(df[['Balance','EstimatedSalary','CreditScore','Age']])
```

```python
y=df['Exited']
```

```python
y
```

```python
x=df.drop(['Exited'],axis=1)
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.3, random_state=42, stratify=y)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```python

GB_model=GradientBoostingClassifier()
```

```python

GB_model.fit(X_train,y_train)
```

```python
from sklearn.model_selection import RandomizedSearchCV
from sklearn.ensemble import GradientBoostingRegressor
from scipy.stats import randint, uniform
```

```python
param_grid = {
    'n_estimators': randint(50, 200),  # Integers between 50 and 199
    'learning_rate': uniform(0.01, 0.2), # Uniform distribution between 0.01 and 0.21
    'max_depth': randint(3, 10),       # Integers between 3 and 9
    'subsample': uniform(0.7, 0.3),    # Uniform distribution between 0.7 and 1.0
    'min_samples_split': randint(2, 11), # Integers between 2 and 10
    'min_samples_leaf': randint(1, 5)   # Integers between 1 and 4
}
```

```python
from sklearn.model_selection import RandomizedSearchCV

random_search = RandomizedSearchCV(
    estimator=GB_model,
    param_distributions=param_grid,
    n_iter=30,  # Only test 30 random combinations
    scoring='f1',
    cv=3,
    n_jobs=-1,
    verbose=2,
    random_state=42
)
```

```python
random_search.fit(X_train, y_train)
```

```python
print("Best Parameters:", random_search.best_params_)
```

```python
gb_final = GradientBoostingClassifier(
    subsample=0.7,
    n_estimators=300,
    min_samples_split=5,
    min_samples_leaf=3,
    max_depth=5,
    learning_rate=0.05,
    random_state=42
)

gb_final.fit(X_train, y_train)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

y_pred = gb_final.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall:", recall_score(y_test, y_pred))
print("F1 Score:", f1_score(y_test, y_pred))
```
