```python
df.head()
```

```python
df=pd.get_dummies(df,columns=['Gender'], prefix='Is',drop_first=True)
```

```python
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
df['Geography'] = label_encoder.fit_transform(df['Geography'])
```

```python
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
df[['scaled_Balance','scaled_EstimatedSalary','scaled_CreditScore','scaled_Age']]=scaler.fit_transform(df[['Balance','EstimatedSalary','CreditScore','Age']])
```

```python
df.head()
```

```python
df.drop(['CreditScore','Age','Balance','EstimatedSalary'],axis=1,inplace=True)
```

```python
x = df.drop('Exited', axis=1)
y = df['Exited']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42, stratify=y)
```

```python
print("Original training class distribution:", Counter(y_train))
print("Original testing class distribution:", Counter(y_test))
```

```python
from imblearn.over_sampling import SMOTE
```

```python
smote = SMOTE(random_state=42)
```

```python
X_resampled, y_resampled = smote.fit_resample(x_train, y_train)
```

```python
print("Resampled training class distribution:", Counter(y_resampled))
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
```

```python

from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
```

```python
model=RandomForestClassifier(
    max_depth= 20,
    max_features= None,
    min_samples_leaf= 1,
    min_samples_split= 2,
    n_estimators= 1000,
    random_state=42,
    criterion= 'entropy')
```

```python
model.fit(X_resampled, y_resampled)
```

```python
y_pred=model.predict(x_test)
```

```python
accuracy_score(y_pred,y_test)
```
